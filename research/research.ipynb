```python
file_path = "data/DM.pdf"
loader = PyPDFLoader(file_path)
data = loader.load()
```

```python
data
```

```python
len(data)
```

```python
question_gen = ""
for page in data:
    question_gen += page.page_content
```

```python
question_gen
```

```python
## Do chunking
from langchain_text_splitters import TokenTextSplitter
```

```python
splitter_ques_gen = TokenTextSplitter( model_name="gpt-3.5-turbo"
                                      , chunk_size=1000
                                      , chunk_overlap=50)
```

```python
chunks_ques_gen = splitter_ques_gen.split_text(question_gen)
```

```python
chunks_ques_gen
```

```python
len(chunks_ques_gen)
```

```python
## chunks_ques_gen is a list of all str elements. We need to convert it to a list of Document objects.
from langchain_core.documents import Document
documents_ques_gen = [Document(page_content=chunk) for chunk in chunks_ques_gen]
```

```python
documents_ques_gen
```

```python
type(documents_ques_gen[0])
```

```python
splitter_ans_gen = TokenTextSplitter( model_name="gpt-3.5-turbo" 
                                    , chunk_size=1000
                                    , chunk_overlap=50)
```

```python
document_answer_gen = splitter_ans_gen.split_documents(documents_ques_gen)
```

```python
document_answer_gen
```

```python
from langchain_openai import ChatOpenAI
```

```python
llm_ques_gen_pipeline = ChatOpenAI(model_name="gpt-3.5-turbo", temperature=0.3)
```

```python
prompt_template = """
You are a helpful assistant for generating multiple choice questions (MCQs) 
based on the text given below
                     
-------
{text}
-------

Create multiple choin type questions based on the above text. Each question should have 4 options and only one correct answer.
Make sure not to loose any important information from the text while creating the questions. The questions should be of high quality and should be able to test the understanding of the text.
                     
Questions:
"""
```

```python
from langchain_core.prompts import PromptTemplate
```

```python
PROMPT_QUESTIONS = PromptTemplate(template=prompt_template, input_variables=["text"])
```

```python
refine_template = ("""
You are an expert in generating multiple choice questions (MCQs) based on the text given below. Your goal is to 
refine the questions generated by the previous model to make them more accurate, relevant, and of higher quality.

We have received some paractice multiple choice questions to a certain extent : {existing_answer}. 
We have the option to modify these questions or to add new ones to improve their quality with
some context below.
-----------------                 
{text}. 
-----------------
Given the new context, refine the existing multiple choice questions to make them more accurate, relevant, and of higher quality. Make sure to retain the core information from the original text while improving the questions.
If the question is not helpful, please provide the original question without any modifications.

Questions:

""")
```

```python
REFINE_PROMPT_QUESTIONS = PromptTemplate(input_variables=["existing_answer", "text"]
                                         ,template=refine_template)
```

```python
from langchain_classic.chains.summarize.chain import load_summarize_chain
```

```python
ques_gen_chain = load_summarize_chain(llm= llm_ques_gen_pipeline
                                      ,chain_type="refine"
                                      ,verbose=True
                                      ,question_prompt = PROMPT_QUESTIONS
                                      ,refine_prompt = REFINE_PROMPT_QUESTIONS)
```

```python
ques = ques_gen_chain.run(documents_ques_gen)
print(ques)
```

```python
from langchain_community.embeddings.openai import OpenAIEmbeddings
```

```python
embeddings = OpenAIEmbeddings()
```

```python
from langchain_community.vectorstores import FAISS
```

```python
vector_store = FAISS.from_documents(document_answer_gen,embeddings)
```

```python
llm_answer_gen = ChatOpenAI(temperature=0.1,model="gpt-3.5-turbo")
```

```python
ques
```

```python
ques_list = ques.split("\n")
```

```python
ques_list
```

```python
for question in ques_list:
    #Save Answers into a file
    with open("question.txt","a") as f:
        f.write(question+"\n")
```

```python
from langchain_classic.chains import RetrievalQA
```

```python
answer_generation_chain = RetrievalQA.from_chain_type(llm=llm_answer_gen,
                                                      chain_type = "stuff",
                                                      retriever = vector_store.as_retriever())
```

```python
for question in ques_list:
    answer = answer_generation_chain.run(question)
    #Save Answers into a file
    with open("answer.txt","a") as f:
        f.write(question+"\n")
        f.write(answer + "\n")
```
